```python
family_tree = {
    "Alice": ["Bob", "Carol", "David"],
    "Bob": ["Emma", "Finn"],
    "Carol": ["Grace", "Henry"],
    "David": ["Iris", "Jack"],
    "Emma": ["Kate", "Leo"],
    "Finn": [],
    "Grace": ["Mia"],
    "Henry": [],
    "Iris": [],
    "Jack": ["Noah"],
    "Kate": [],
    "Leo": [],
    "Mia": [],
    "Noah": []
}
```

```python
def depth_first_traversal(tree, root):
    next_node = [root, ]
    while len(next_node) > 0:
        current_node = next_node.pop()
        print(current_node)
        next_node.extend(tree[current_node][::-1])  # Add children in reverse order for correct traversal

depth_first_traversal(family_tree, "Alice")
```

```python
def breadth_first_traversal(tree, root):
    next_node = [root, ]
    while len(next_node) > 0:
        current_node = next_node.pop(0)
        print(current_node)
        next_node.extend(tree[current_node])  # Add children in order for correct traversal

breadth_first_traversal(family_tree, "Alice")
```
