```python
def current_candidate_is_valid(current_sub_candidate, starting_numbers):
    if len(current_sub_candidate) == 0:
        return True

    current_index = len(current_sub_candidate) - 1

    # Elk getal mag maar één keer voorkomen.
    if current_sub_candidate[-1] in current_sub_candidate[:-1]:
        return False

    # Aangrenzende getallen mogen niet opeenvolgend zijn.
    if current_index > 0 and abs(current_sub_candidate[-1] - current_sub_candidate[-2]) == 1:
        return False

    # Controleer of een vooringevuld getal niet gewijzigd werd.
    if starting_numbers[current_index] is not None and current_sub_candidate[-1] != starting_numbers[current_index]:
        return False
    
    return True

def generate_correct_solutions(starting_numbers, current_sub_candidate = None, solutions = None, size = 7):
    if current_sub_candidate is None:
        current_sub_candidate = []
    if solutions is None:
        solutions = []

    if not current_candidate_is_valid(current_sub_candidate, starting_numbers):
        return # Backtracking

    if len(current_sub_candidate) == size:
        solutions.append(current_sub_candidate)
        return
    
    for i in range(1, size + 1):
        next_sub_candidate = current_sub_candidate + [i, ]
        generate_correct_solutions(starting_numbers, next_sub_candidate, solutions, size)

    return solutions

solutions = generate_correct_solutions([None, None, 2, None, None, 1, 6])
print(len(solutions))
print(solutions)
```
